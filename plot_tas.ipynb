```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
import itertools

from matplotlib.colors import BoundaryNorm
from global_land_mask import is_land

from KSCALE.read_data.read_data_catalog import load_data_global, load_data_regional
from KSCALE.p_config import regions
```

```python
region='Africa'
variable = 'tas'
zoom_low = 7
zoom_high = 8

year = 2020
months = np.arange(1, 12+1, 1)

ts = '1M'  # resampling timescale

lat_range = regions[region][0]
lon_range = regions[region][1]
lat_min = lat_range[0]
lat_max = lat_range[1]
lon_min = lon_range[0]
lon_max = lon_range[1]
```

```python
data_gal_zlow = load_data_global(resolution='n1280', driving='GAL9', zoom=zoom_low, variable=variable,
                                 lat_range=lat_range, lon_range=lon_range, year=year, timescale=ts)
data_ral_zlow = load_data_global(resolution='n2560', driving='RAL3p3', zoom=zoom_low, variable=variable,
                                 lat_range=lat_range, lon_range=lon_range, year=year, timescale=ts)
data_ctc_zlow = load_data_regional(region='CTC', resolution='km4p4', physics='RAL3P3', driving='GAL9', reso_nest='n1280', zoom=zoom_low, variable=variable,
                                 lat_range=lat_range, lon_range=lon_range, year=year, timescale=ts)
#data_gal_zhigh = load_data_global(resolution='n1280', driving='GAL9', zoom=zoom_high, variable=variable,
#                                lat_range=lat_range, lon_range=lon_range, year=year, timescale=ts)
#data_ral_zhigh = load_data_global(resolution='n2560', driving='RAL3p3', zoom=zoom_high, variable=variable,
#                                 lat_range=lat_range, lon_range=lon_range, year=year, timescale=ts)
data_lam_zlow = load_data_regional(region=region, resolution='km4p4', physics='RAL3P3', driving='GAL9', reso_nest='n1280', zoom=zoom_low, variable=variable,
                                 lat_range=lat_range, lon_range=lon_range, year=year, timescale=ts)
```

```python
#~ Mask ocean

data_gal_zlow_tmean = data_gal_zlow_tmean.where(land_mask_zlow)
data_ral_zlow_tmean = data_ral_zlow_tmean.where(land_mask_zlow)
data_lam_zlow_tmean = data_lam_zlow_tmean.where(land_mask_zlow)

#data_gal_zhigh_tmean = data_gal_zhigh_tmean.where(land_mask_zhigh)
#data_ral_zhigh_tmean = data_ral_zhigh_tmean.where(land_mask_zhigh)
```

```python
data_min = min(abs(data_gal_zlow_tmean.min()), abs(data_ral_zlow_tmean.min()), abs(data_lam_zlow_tmean.min())).values
data_max = max(abs(data_gal_zlow_tmean.max()), abs(data_ral_zlow_tmean.max()), abs(data_lam_zlow_tmean.max())).values
data_min, data_max
```

```python
diff_min = min(diff_zlow.min(), diff_gal_lam_zlow.min(), diff_ral_lam_zlow.min()).values
diff_max = min(abs(diff_zlow.max()), abs(diff_gal_lam_zlow.max()), abs(diff_ral_lam_zlow.max())).values
diff_lim = max(abs(diff_min), abs(diff_max))
diff_lim
```

```python
cmap = plt.get_cmap('YlOrRd')
bounds = np.linspace(20, 35, 16)
norm = BoundaryNorm(bounds, cmap.N, extend='both')
cmap_diff = plt.get_cmap('RdBu_r')
bounds_diff = np.linspace(-10, 10, 11)
norm_diff = BoundaryNorm(bounds_diff, cmap.N, extend='both')

fig, ax = plt.subplots(ncols=4, nrows=4)
fig_diff, ax_diff = plt.subplots(ncols=4, nrows=4)

for i, (seas, mths) in enumerate(months_all.items()):
    lmonths = [imths[m] for m in mths]
    lmonths = list(itertools.chain(*lmonths))

    ax_ = ax[i]
    ax_diff_ = ax_diff[i]

    data_gal_zlow_ms = data_gal_zlow.isel(time=lmonths).mean(dim='time').where(land_mask_zlow) - 273.15
    data_ral_zlow_ms = data_ral_zlow.isel(time=lmonths).mean(dim='time').where(land_mask_zlow) - 273.15
    data_ctc_zlow_ms = data_ctc_zlow.isel(time=lmonths).mean(dim='time').where(land_mask_zlow) - 273.15
    data_lam_zlow_ms = data_lam_zlow.isel(time=lmonths).mean(dim='time').where(land_mask_zlow) - 273.15

    diff_ral_zlow_ms = data_ral_zlow_ms - data_gal_zlow_ms
    diff_ctc_zlow_ms = data_ctc_zlow_ms - data_gal_zlow_ms
    diff_lam_gal_zlow_ms = data_lam_zlow_ms - data_gal_zlow_ms
    diff_lam_ral_zlow_ms = data_lam_zlow_ms - data_ral_zlow_ms

    p = data_gal_zlow_ms.plot(ax=ax_[0], norm=norm, cmap=cmap, add_colorbar=False)
    data_ctc_zlow_ms.plot(ax=ax_[1], norm=norm, cmap=cmap, add_colorbar=False)
    data_ral_zlow_ms.plot(ax=ax_[2], norm=norm, cmap=cmap, add_colorbar=False)
    data_lam_zlow_ms.plot(ax=ax_[3], norm=norm, cmap=cmap, add_colorbar=False)

    p_diff = diff_ral_zlow_ms.plot(ax=ax_diff_[0], norm=norm_diff, cmap=cmap_diff, add_colorbar=False)
    diff_ctc_zlow_ms.plot(ax=ax_diff_[1], norm=norm_diff, cmap=cmap_diff, add_colorbar=False)
    diff_lam_gal_zlow_ms.plot(ax=ax_diff_[2], norm=norm_diff, cmap=cmap_diff, add_colorbar=False)
    diff_lam_ral_zlow_ms.plot(ax=ax_diff_[3], norm=norm_diff, cmap=cmap_diff, add_colorbar=False)

    ax_[0].text(0.025, 0.025, '%.2f C'%data_gal_zlow_ms.mean(), transform=ax_[0].transAxes)
    ax_[1].text(0.025, 0.025, '%.2f C'%data_ral_zlow_ms.mean(), transform=ax_[1].transAxes)
    ax_[2].text(0.025, 0.025, '%.2f C'%data_ctc_zlow_ms.mean(), transform=ax_[2].transAxes)
    ax_[3].text(0.025, 0.025, '%.2f C'%data_lam_zlow_ms.mean(), transform=ax_[3].transAxes)

    ax_[0].set_xlabel('')
    ax_[1].set_xlabel('')
    ax_[2].set_xlabel('')
    ax_[3].set_xlabel('')
    ax_diff_[0].set_xlabel('')
    ax_diff_[1].set_xlabel('')
    ax_diff_[2].set_xlabel('')
    ax_diff_[3].set_xlabel('')

    ax_[0].set_ylabel(seas)
    ax_[1].set_ylabel('')
    ax_[2].set_ylabel('')
    ax_[3].set_ylabel('')

    ax_diff_[0].set_ylabel(seas)
    ax_diff_[0].grid(True, alpha=0.2)
    ax_diff_[1].grid(True, alpha=0.2)
    ax_diff_[2].grid(True, alpha=0.2)
    ax_diff_[3].grid(True, alpha=0.2)

    ax_[0].grid(True, alpha=0.2)
    ax_[1].grid(True, alpha=0.2)
    ax_[2].grid(True, alpha=0.2)
    ax_[3].grid(True, alpha=0.2)

    ax_diff_[1].set_ylabel('')
    ax_diff_[2].set_ylabel('')
    ax_diff_[3].set_ylabel('')

    if i == 0:
        ax_[0].set_title('GAL9')
        ax_[1].set_title('RAL3 (n2560)')
        ax_[2].set_title('CTC (km4p4)')
        ax_[3].set_title('RAL3 (km4p4)')
        ax_diff_[0].set_title('RAL3_n2560$-$CoMA9')
        ax_diff_[1].set_title('CTC$-$CoMA9')
        ax_diff_[2].set_title('RAL3_km4p4$-$CoMA9')
        ax_diff_[3].set_title('RAL3_km4p4$-$RAL3_n2560')
    else:
        ax_[0].set_title('')
        ax_[1].set_title('')
        ax_[2].set_title('')
        ax_[3].set_title('')
        ax_diff_[0].set_title('')
        ax_diff_[1].set_title('')
        ax_diff_[2].set_title('')
        ax_diff_[3].set_title('')

cb = plt.colorbar(p, norm=norm, cmap=cmap, ax=ax[:], orientation='horizontal', label='Tas [C]', shrink=0.5, pad=0.1)
cb_diff = plt.colorbar(p_diff, norm=norm_diff, cmap=cmap_diff, ax=ax_diff[:], orientation='horizontal', label='$\\Delta$ Tas [C]', shrink=0.5, pad=0.075)

#data_ral_zhigh_ms.plot(ax=ax[3], norm=norm, cmap=cmap, add_colorbar=False)
#data_lam_zlow_ms.plot(ax=ax[4], norm=norm, cmap=cmap, add_colorbar=False)

fig.set_size_inches(12., 9.)
fig_diff.set_size_inches(12., 9.)
```
